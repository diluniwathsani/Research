```python
# 1. Load your data from Notebook 8
df = pd.read_csv('final_expanded_requirements.csv')

def generate_professional_artifacts(row):
    req = row['requirement_sentence']
    
    # --- STEP 1: DEFINE THE EPIC (Based on SVM Category) ---
    if row['Security'] == 1:
        epic = "Data Security & Access Management"
        feature = "System Hardening"
    elif row['Reliability'] == 1:
        epic = "System Resilience"
        feature = "Availability & Recovery"
    elif row['NFR_boolean'] == 1:
        epic = "System Quality Attributes"
        feature = "Performance Standards"
    else:
        epic = "Functional Operations"
        feature = "Core Business Logic"

    # --- STEP 2: GENERATE THE USER STORY ---
    # We remove "The system shall" to make it sound like a story
    clean_req = req.replace("The system shall", "").replace("the system shall", "").strip()
    # Capitalize the first letter of the cleaned requirement
    clean_req = clean_req[0].upper() + clean_req[1:] if clean_req else clean_req
    
    user_story = f"As a System User, I want {clean_req} so that the system fulfills its business objectives."

    # --- STEP 3: GENERATE ACCEPTANCE CRITERIA ---
    ac = f"Given the system is in a normal state, When the requirement '{clean_req}' is triggered, Then the system must successfully execute the action and log the result."

    return pd.Series([epic, feature, user_story, ac])
```

```python
#2. Apply the logic to create the columns
print("Generating artifacts using Template Logic...")
df[['Epic', 'Feature', 'User_Story', 'Acceptance_Criteria']] = df.apply(generate_professional_artifacts, axis=1)
```

```python
# 3. Save the final result
df.to_csv('final_ba_artifacts.csv', index=False)
```
